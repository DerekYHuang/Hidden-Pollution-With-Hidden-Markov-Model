```python
import pandas as pd
import numpy as np
```

```python
# Load the dataset. Note: Using semicolon as delimiter and specifying decimal comma
df = pd.read_csv('../data/raw/daily_precip_san_diego.csv', sep=',')
df.head()
```

```python
# Drop rows where station != sd
df = df[df["STATION_NAME"] == "SAN DIEGO INTERNATIONAL AIRPORT CA US"]
df = df[["HPCP", "DATE"]]
df.head()
```

```python
df["DATE"] = pd.to_datetime(df["DATE"], format="%Y%m%d %H:%M").dt.strftime("%Y-%m-%d")
df["HPCP"] = df["HPCP"].apply(lambda x: 1 if x > 0 else 0)

df = df.reset_index(drop=True)

df.head()
```

```python
df.to_csv("../data/processed/precip_data_cleaned.csv",header=True )
```
