```python
import pandas as pd
import numpy as np
```

```python
# Load the dataset. Note: Using semicolon as delimiter and specifying decimal comma
df = pd.read_csv('../data/raw/AirQualityUCI.csv', sep=';', decimal=',')

# Remove empty columns (columns with all NaN or empty strings)
df = df.dropna(axis=1, how='all')
df = df.loc[:, (df != '').any(axis=0)]
df.head()
```

```python
# Clean column names
column_mapping = {
    # Pollutant concentrations
    'CO(GT)': 'CO',
    'NO2(GT)': 'NO2',

    # Metal-oxide sensor channels
    'PT08.S5(O3)': 'O3',
}
df = df.rename(columns=column_mapping)

# Keep CO, NO2, O3 
    # Used to later calcualte AQI
    # Based on 3/5 EPA's five major air pollutants for AQI (missing SO2 & PM10/2.5)
df = df[['CO', 'NO2', 'O3']]
df = df.dropna() # Drop rows where any column is null

df.head()
```

```python
df.to_csv("../data/processed/airquality_cleaned.csv",header=True )
```
