```python
import pandas as pd
import os

# Load the dataset
df = pd.read_csv('../data/raw/daily_aqi_by_county_2013.csv')  # Replace with your actual file path

print("Original dataset shape:", df.shape)

# Filter for California, San Diego only
df_san_diego = df[(df['State Name'] == 'California') & 
                   (df['county Name'] == 'San Diego')].copy()

print("\n" + "="*60)
print("FILTERING FOR SAN DIEGO, CALIFORNIA")
print("="*60)
print(f"San Diego dataset shape: {df_san_diego.shape}")
print(f"Records found: {len(df_san_diego)}")
print(f"Percentage of original data: {(len(df_san_diego)/len(df)*100):.2f}%")

# Check if any data was found
if len(df_san_diego) == 0:
    print("\n⚠️  WARNING: No data found for San Diego, California!")
    print("\nChecking available California counties:")
    ca_counties = df[df['State Name'] == 'California']['county Name'].unique()
    print(f"Found {len(ca_counties)} counties in California:")
    print(sorted(ca_counties))
else:
    # Display summary statistics
    print("\n" + "="*60)
    print("SAN DIEGO DATA SUMMARY")
    print("="*60)
    
    # Date range
    df_san_diego['Date'] = pd.to_datetime(df_san_diego['Date'])
    print(f"\nDate range: {df_san_diego['Date'].min()} to {df_san_diego['Date'].max()}")
    
    # AQI statistics
    print(f"\nAQI Statistics:")
    print(f"  Mean AQI: {df_san_diego['AQI'].mean():.2f}")
    print(f"  Min AQI: {df_san_diego['AQI'].min()}")
    print(f"  Max AQI: {df_san_diego['AQI'].max()}")
    print(f"  Median AQI: {df_san_diego['AQI'].median():.2f}")
    
    # Category distribution
    print(f"\nAir Quality Category Distribution:")
    category_counts = df_san_diego['Category'].value_counts()
    for category, count in category_counts.items():
        pct = (count / len(df_san_diego)) * 100
        print(f"  {category}: {count} ({pct:.1f}%)")
    
    # Defining Parameter distribution
    print(f"\nDefining Parameter Distribution:")
    parameter_counts = df_san_diego['Defining Parameter'].value_counts()
    for param, count in parameter_counts.items():
        pct = (count / len(df_san_diego)) * 100
        print(f"  {param}: {count} ({pct:.1f}%)")
    
    # Number of unique sites
    print(f"\nUnique monitoring sites: {df_san_diego['Defining Site'].nunique()}")
    print(f"Site IDs: {df_san_diego['Defining Site'].unique()}")
    
    # Display first few rows
    print("\n" + "="*60)
    print("FIRST 10 ROWS OF SAN DIEGO DATA")
    print("="*60)
    print(df_san_diego.head(10))
    
    print("✓ San Diego dataframe created: 'df_san_diego'")
```

```python
# Filter to keep only Date and AQI columns
df_date_aqi = df_san_diego[['Date', 'AQI']].copy()

print("Filtered dataframe shape:", df_date_aqi.shape)
print("\n" + "="*60)
print("DATE AND AQI DATA SUMMARY")
print("="*60)

# Ensure Date is in datetime format
df_date_aqi['Date'] = pd.to_datetime(df_date_aqi['Date'])

print(f"\nTotal records: {len(df_date_aqi)}")
print(f"Date range: {df_date_aqi['Date'].min()} to {df_date_aqi['Date'].max()}")
print(f"\nAQI Statistics:")
print(f"  Mean: {df_date_aqi['AQI'].mean():.2f}")
print(f"  Median: {df_date_aqi['AQI'].median():.2f}")
print(f"  Min: {df_date_aqi['AQI'].min()}")
print(f"  Max: {df_date_aqi['AQI'].max()}")
print(f"  Std Dev: {df_date_aqi['AQI'].std():.2f}")

# Display first 10 rows
print("\n" + "="*60)
print("FIRST 10 ROWS")
print("="*60)
print(df_date_aqi.head(10))

# Save to CSV in data/processed folder
df_date_aqi.to_csv("../data/processed/aqi_cleaned.csv", header=True, index=False)
```
