```python
import pandas as pd
# Load the dataset
df = pd.read_csv('../data/raw/daily_aqi_by_county_2013.csv')  # Replace with your actual file path
df.head()
```

```python
# Filter for California, San Diego only
df_san_diego = df[(df['State Name'] == 'California') & (df['county Name'] == 'San Diego')].copy()
df_date_aqi = df_san_diego[['Date', 'AQI']].copy()
df_date_aqi['Date'] = pd.to_datetime(df_date_aqi['Date'])

df_date_aqi.head()
```

```python
df_date_aqi.to_csv("../data/processed/aqi_cleaned.csv", header=True, index=False)
```
